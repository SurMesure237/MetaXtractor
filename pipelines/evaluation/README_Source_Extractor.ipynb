# README Source Extractor
Extract entries with README.md sources from JSON data

```python
import json

# Configuration
JSON_PATH = "test1.json"  # Update this path
OUTPUT_PATH = "readme_entries.json"
```

```python
def extract_readme_entries(data):
    """Extract entries with README.md sources"""
    readme_entries = []
    
    for field_name, field_data in data.items():
        if isinstance(field_data, list):
            for entry in field_data:
                if isinstance(entry, dict) and 'source' in entry:
                    source = entry['source']
                    if isinstance(source, str) and source.endswith('README.md'):
                        readme_entries.append({
                            'field': field_name,
                            'entry': entry
                        })
    
    return readme_entries

# Load and process data
with open(JSON_PATH, 'r') as f:
    data = json.load(f)

readme_entries = extract_readme_entries(data)

print(f"Found {len(readme_entries)} entries with README.md sources:")
for entry in readme_entries:
    print(f"- Field: {entry['field']}")
    print(f"  Source: {entry['entry']['source']}")
    print(f"  Value: {entry['entry'].get('result', {}).get('value', 'N/A')}")
    print()
```

```python
# Export results
with open(OUTPUT_PATH, 'w') as f:
    json.dump(readme_entries, f, indent=2)

print(f"Exported {len(readme_entries)} entries to {OUTPUT_PATH}")
```
